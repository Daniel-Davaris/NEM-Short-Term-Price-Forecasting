# 3.4 Save the final model

**Input:** 96 saved residual stackers.

**Output:** `final_params.joblib`, containing the models required to produce one price prediction per horizon.

```python
# Step 1 - Load the stackers

import os
import sys
from datetime import datetime, timezone

import joblib
import pandas as pd

sys.path.append(os.path.abspath("../../")) ; from EPF import variables

HORIZONS = range(1, variables.HORIZON_COUNT+1)
stackers = [joblib.load(variables.RESIDUAL_STACKERS_DIR / f"h{horizon:02d}_residual_stacker.joblib") for horizon in HORIZONS]
diagnostics = pd.read_csv(variables.RESIDUAL_STACKER_DIAGNOSTICS_PATH)
```

```python
# Step 2 - Save the deployable model

final_model = {
    "schema_version": 5,
    "approach": "residual_stacking",
    "created_utc": datetime.now(timezone.utc).isoformat(),
    "target_region": variables.TARGET_REGION,
    "stackers": stackers,
    "validation_diagnostics": diagnostics.to_dict("records"),
}

joblib.dump(final_model, variables.FINAL_PARAMS_PATH)
print(variables.FINAL_PARAMS_PATH)
```
